
# 🔗 Merge Club Features into Player Dataset (Using Local Paths)

This notebook merges your three main datasets:

- `clubs_dataset.csv`: Base dataset with club names and core metadata
- `african_clubs_features_dataset.csv`: Club-level enrichment (tier, exposure, etc.)
- `players_dataset.csv`: Player dataset to be enriched with club + league features

We will:
1. Load and inspect all files
2. Merge club features into clubs dataset
3. Merge enriched clubs into player dataset
4. Export a new `players_dataset_enriched.csv` file


## 📥 Step 1: Load All Datasets

```python
import pandas as pd

# Load datasets using your full paths
clubs_df = pd.read_csv(r"C:/Users/nadza/OneDrive/Desktop/Tactic_Sense_Dataset/clubs_dataset.csv")
features_df = pd.read_csv(r"C:/Users/nadza/OneDrive/Desktop/Tactic_Sense_Dataset/african_clubs_features_dataset.csv")
players_df = pd.read_csv(r"C:/Users/nadza/OneDrive/Desktop/Tactic_Sense_Dataset/players_dataset.csv")

# Preview clubs dataset
clubs_df.head()

```

# Merge club-level features into base club data
clubs_enriched = clubs_df.merge(features_df, how="left", on="club name")

# Preview result
clubs_enriched.head()



```python
print(clubs_df.columns.tolist())
print(features_df.columns.tolist())
```

```python
# Merge club-level features into clubs
clubs_enriched = clubs_df.merge(features_df, how="left", left_on="club name", right_on="club_name")

# Preview result
clubs_enriched.head()
```

## 👥 Step 3: Merge Clubs into Players Dataset

```python
print(players_df.columns.tolist())
```

```python
# Merge enriched club features into player dataset
players_enriched = players_df.merge(
    clubs_enriched, 
    how="left", 
    left_on="club", 
    right_on="club name"
)

# Fill missing values safely
players_enriched["league_strength"] = players_enriched["league_strength"].fillna(1)
players_enriched["club_tier"] = players_enriched["club_tier"].fillna(1)
players_enriched["league_exposure"] = players_enriched["league_exposure"].fillna("Low")
players_enriched["international_exposure"] = players_enriched["international_exposure"].fillna("No")

# Preview
players_enriched.head()


```

## 💾 Step 4: Export Final Enriched Player Dataset

```python
# Export the final enriched player dataset to your folder
players_enriched.to_csv(r"C:/Users/nadza/OneDrive/Desktop/Tactic_Sense_Dataset/players_dataset_enriched.csv", index=False)

print("✅ Exported successfully to: C:/Users/nadza/OneDrive/Desktop/Tactic_Sense_Dataset/players_dataset_enriched.csv")
```
