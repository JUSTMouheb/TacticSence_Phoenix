```python
# Convert data types without handling missing values
def preprocess_dataset(df, numeric_cols=None, categorical_cols=None, boolean_cols=None):
    """
    Preprocess a dataset by ensuring correct data types
    
    Parameters:
    - df: DataFrame to preprocess
    - numeric_cols: List of columns to convert to numeric
    - categorical_cols: List of columns to convert to categorical
    - boolean_cols: List of columns to convert to boolean
    
    Returns:
    - Preprocessed DataFrame
    """
    # Make a copy to avoid modifying original
    processed_df = df.copy()
    
    # Convert numeric columns
    if numeric_cols:
        for col in numeric_cols:
            if col in processed_df.columns:
                processed_df[col] = pd.to_numeric(processed_df[col], errors='coerce')
    
    # Convert categorical columns
    if categorical_cols:
        for col in categorical_cols:
            if col in processed_df.columns:
                processed_df[col] = processed_df[col].astype(str)
    
    # Convert boolean columns
    if boolean_cols:
        for col in boolean_cols:
            if col in processed_df.columns:
                # Convert various representations to boolean
                processed_df[col] = processed_df[col].map({
                    True: True, 'True': True, 'true': True, 1: True, '1': True,
                    False: False, 'False': False, 'false': False, 0: False, '0': False
                })
    
    return processed_df
```

```python
def normalize_value(value, min_value, max_value):
    """Normalize a value to the range 0-100"""
    if pd.isna(value) or max_value == min_value:
        return 50  # Default middle value if no range or missing value
    
    normalized = 100 * (value - min_value) / (max_value - min_value)
    return max(0, min(100, normalized))

def get_verified_bonus(verified_status):
    """Return verification bonus score (0-100) based on verified status"""
    if verified_status is True or verified_status == 'True' or verified_status == 'true' or verified_status == 'Yes' or verified_status == 'yes':
        return 100
    return 0
```

```python
def calculate_player_quality(player, players_df):
    """Calculate quality score for a player"""
    # Market value (normalized)
    max_value = players_df['market_value_eur'].max()
    min_value = players_df['market_value_eur'].min()
    normalized_value = normalize_value(player['market_value_eur'], min_value, max_value) if pd.notna(player['market_value_eur']) else 50
    
    # Fitness level (0-10 scale to 0-100)
    fitness_level = player['fitness_level'] * 10 if pd.notna(player['fitness_level']) else 50
    
    # Profile rating (0-10 scale to 0-100)
    profile_rating = player['profile_rating'] * 10 if pd.notna(player['profile_rating']) else 50
    
    # Goals and assists contribution
    goals = player['goals'] if pd.notna(player['goals']) else 0
    assists = player['assists'] if pd.notna(player['assists']) else 0
    contribution = min(100, (goals + assists) * 5)  # Cap at 100
    
    # Scouting score (0-10 scale to 0-100)
    scouting_score = player['scouting_score'] * 10 if pd.notna(player['scouting_score']) else 50
    
    # Verified status bonus
    verified_bonus = get_verified_bonus(player['verified_status'])
    
    # Calculate weighted quality score
    quality_score = (0.3 * normalized_value) + (0.2 * fitness_level) + (0.15 * profile_rating) + \
                    (0.15 * contribution) + (0.1 * scouting_score) + (0.1 * verified_bonus)
    
    return max(0, min(100, quality_score))

def calculate_club_quality(club, clubs_df):
    """Calculate quality score for a club"""
    # Ranking (inverse and normalize)
    max_rank = clubs_df['ranking'].max()
    min_rank = clubs_df['ranking'].min()
    # Invert ranking so lower is better
    normalized_rank = normalize_value(max_rank - club['ranking'] + min_rank, min_rank, max_rank) if pd.notna(club['ranking']) else 50
    
    # Market budget (normalized)
    max_budget = clubs_df['market_budget'].max()
    min_budget = clubs_df['market_budget'].min()
    normalized_budget = normalize_value(club['market_budget'], min_budget, max_budget) if pd.notna(club['market_budget']) else 50
    
    # Profile rating (0-10 scale to 0-100)
    profile_rating = club['profile_rating'] * 10 if pd.notna(club['profile_rating']) else 50
    
    # Avg transfer fee (normalized)
    max_fee = clubs_df['avg_transfer_fee_paid'].max()
    min_fee = clubs_df['avg_transfer_fee_paid'].min()
    normalized_fee = normalize_value(club['avg_transfer_fee_paid'], min_fee, max_fee) if pd.notna(club['avg_transfer_fee_paid']) else 50
    
    # Verified status bonus
    verified_bonus = get_verified_bonus(club['verified_status'])
    
    # Calculate weighted quality score
    quality_score = (0.3 * normalized_rank) + (0.25 * normalized_budget) + \
                    (0.2 * profile_rating) + (0.15 * normalized_fee) + (0.1 * verified_bonus)
    
    return max(0, min(100, quality_score))

def calculate_player_agent_quality(agent, agents_df):
    """Calculate quality score for a player agent"""
    # Success rate (0-100 scale)
    success_rate = agent['success_rate'] if pd.notna(agent['success_rate']) else 50
    
    # Total clients (assume 20+ clients is 100%)
    total_clients = min(100, max(0, agent['total_clients'] * 5)) if pd.notna(agent['total_clients']) else 0
    
    # Average contract value (normalized)
    max_value = agents_df['avg_contract_value'].max()
    min_value = agents_df['avg_contract_value'].min()
    normalized_value = normalize_value(agent['avg_contract_value'], min_value, max_value) if pd.notna(agent['avg_contract_value']) else 50
    
    # Profile rating (0-10 scale to 0-100)
    profile_rating = agent['profile_rating'] * 10 if pd.notna(agent['profile_rating']) else 50
    
    # Years experience (assume 20+ years is 100%)
    years_experience = min(100, max(0, agent['years_experience'] * 5)) if pd.notna(agent['years_experience']) else 0
    
    # Verified status bonus
    verified_bonus = get_verified_bonus(agent['verified_status'])
    
    # Calculate weighted quality score
    quality_score = (0.25 * success_rate) + (0.2 * total_clients) + \
                    (0.2 * normalized_value) + (0.15 * profile_rating) + \
                    (0.1 * years_experience) + (0.1 * verified_bonus)
    
    return max(0, min(100, quality_score))

# Recruiting Agent quality calculation
def calculate_recruiting_agent_quality(agent, agents_df):
    """Calculate quality score for a recruiting agent"""
    
    # Performance score (0-10 scale to 0-100)
    performance_score = min(100, max(0, agent['Performance_Score'] * 10)) if pd.notna(agent['Performance_Score']) else 50
    
    # Scale successful signings (assume 20+ signings is 100%)
    successful_signings = min(100, max(0, agent['successful_signings'] * 5)) if pd.notna(agent['successful_signings']) else 0
    
    # Scale experience years (assume 20+ years is 100%)
    experience_years = min(100, max(0, agent['experience_years'] * 5)) if pd.notna(agent['experience_years']) else 0
    
    # Positions breadth (number of preferred positions)
    preferred_positions = agent['preferred_positions'] if pd.notna(agent['preferred_positions']) else ''
    positions_breadth = min(100, len(preferred_positions.split(',')) * 20) if preferred_positions else 0
    
    # Verified status bonus
    verified_bonus = get_verified_bonus(agent['verified_status'])
    
    # Calculate weighted quality score
    quality_score = (0.3 * performance_score) + \
                     (0.25 * successful_signings) + \
                     (0.2 * experience_years) + \
                     (0.15 * verified_bonus) + \
                     (0.1 * positions_breadth)
    
    return max(0, min(100, quality_score))

# Manager quality calculation
def calculate_manager_quality(manager, managers_df):
    """Calculate quality score for a manager or staff member"""
    
    # Scale experience years (assume 20+ years is 100%)
    years_experience = min(100, max(0, manager['years_experience'] * 5)) if pd.notna(manager['years_experience']) else 0
    
    # Get profile rating
    profile_rating = min(100, max(0, manager['profile_rating'] * 10)) if pd.notna(manager['profile_rating']) else 0
    
    # Notable achievements score (simplified)
    notable_achievements = manager['notable_achievements'] if pd.notna(manager['notable_achievements']) else ''
    achievements_score = min(100, len(notable_achievements.split(',')) * 25) if notable_achievements else 0
    
    # Certifications score (simplified)
    certifications = manager['certifications'] if pd.notna(manager['certifications']) else ''
    certifications_score = min(100, len(certifications.split(',')) * 20) if certifications else 0
    
    # Availability bonus (small bonus if currently available)
    availability_bonus = 10 if manager.get('availability_status', False) else 0
    
    # Calculate weighted quality score
    quality_score = (0.3 * years_experience) + \
                     (0.25 * profile_rating) + \
                     (0.25 * achievements_score) + \
                     (0.1 * certifications_score) + \
                     (0.1 * availability_bonus)
    
    return max(0, min(100, quality_score))

# Sporting Management Agency quality calculation
def calculate_sporting_mgmt_quality(agency, agencies_df):
    """Calculate quality score for a sporting management agency"""
    
    # Top clients prestige (simplified - count number of clients)
    top_clients = agency['top_clients'] if pd.notna(agency['top_clients']) else ''
    clients_prestige = min(100, len(top_clients.split(',')) * 15) if top_clients else 0
    
    # Services breadth (number of services offered)
    services = agency['services_offered'] if pd.notna(agency['services_offered']) else ''
    services_breadth = min(100, len(services.split(',')) * 20) if services else 0
    
    # Regional influence (simplified)
    region = agency['region'] if pd.notna(agency['region']) else ''
    regional_influence = 80 if region in ['Europe', 'North America'] else 60  # Simplified
    
    # Calculate weighted quality score
    quality_score = (0.3 * clients_prestige) + \
                     (0.3 * services_breadth) + \
                     (0.2 * regional_influence) + \
                     (0.2 * 0)  # No verified status in dataset, so 0 bonus
    
    return max(0, min(100, quality_score))

# Fitness Club quality calculation
def calculate_fitness_club_quality(club, clubs_df):
    """Calculate quality score for a fitness club"""
    
    # Certified trainers (assume 10+ trainers is 100%)
    certified_trainers = min(100, max(0, club['certified_trainers'] * 10)) if pd.notna(club['certified_trainers']) else 0
    
    # Affiliated players (assume 20+ affiliated players is 100%)
    affiliated_players = min(100, max(0, club['affiliated_players'] * 5)) if pd.notna(club['affiliated_players']) else 0
    
    # Services breadth (number of services offered)
    services = club['services_offered'] if pd.notna(club['services_offered']) else ''
    services_breadth = min(100, len(services.split(',')) * 20) if services else 0
    
    # Verified status bonus
    verified_status = club['verified_status'] if pd.notna(club['verified_status']) else False
    verified_bonus = 15 if verified_status == 'Yes' or verified_status is True else 0
    
    # Calculate weighted quality score
    quality_score = (0.3 * certified_trainers) + \
                     (0.25 * verified_bonus) + \
                     (0.25 * affiliated_players) + \
                     (0.2 * services_breadth)
    
    return max(0, min(100, quality_score))

# Equipment Supplier quality calculation
def calculate_equipment_supplier_quality(supplier, suppliers_df):
    """Calculate quality score for an equipment supplier"""
    
    # Price tier (higher price = higher quality, normalized)
    max_price = suppliers_df['price'].max()
    min_price = suppliers_df['price'].min()
    normalized_price = normalize_value(supplier['price'], min_price, max_price) if pd.notna(supplier['price']) else 50
    
    # Affiliated clients prestige (simplified - count number of clients)
    affiliated_clients = supplier['affiliated_clients'] if pd.notna(supplier['affiliated_clients']) else ''
    clients_prestige = min(100, len(affiliated_clients.split(',')) * 10) if affiliated_clients else 0
    
    # Product type breadth (simplified - based on product_type field)
    product_type = supplier['product_type'] if pd.notna(supplier['product_type']) else ''
    product_breadth = min(100, len(product_type.split(',')) * 25) if product_type else 50
    
    # Location prestige (simplified)
    location = supplier['location'] if pd.notna(supplier['location']) else ''
    location_prestige = 90 if location in ['UK', 'US', 'Germany', 'France', 'Italy', 'Spain'] else 70
    
    # Calculate weighted quality score
    quality_score = (0.3 * normalized_price) + \
                     (0.3 * clients_prestige) + \
                     (0.2 * product_breadth) + \
                     (0.2 * location_prestige)
    
    return max(0, min(100, quality_score))

# Sports Clothing Brand quality calculation
# Sports Clothing Brand quality calculation (continued)
def calculate_clothing_brand_quality(brand, brands_df):
    """Calculate quality score for a sports clothing brand"""
    
    # Annual revenue (simplified - parse from string and normalize)
    def extract_revenue(revenue_str):
        if not pd.notna(revenue_str):
            return 50
        try:
            # Extract first number from string
            import re
            matches = re.findall(r'\d+', revenue_str)
            if matches:
                return int(matches[0])
            return 50
        except:
            return 50
    
    revenues = brands_df['annual_revenue'].apply(extract_revenue)
    max_revenue = revenues.max()
    min_revenue = revenues.min()
    brand_revenue = extract_revenue(brand['annual_revenue'])
    normalized_revenue = normalize_value(brand_revenue, min_revenue, max_revenue)
    
    # Sponsorship deals prestige (simplified - count number of deals)
    sponsorship_deals = brand['sponsorship_deals'] if pd.notna(brand['sponsorship_deals']) else ''
    deals_prestige = min(100, len(sponsorship_deals.split(',')) * 15) if sponsorship_deals else 0
    
    # Product types breadth (number of product types)
    product_types = brand['product_types'] if pd.notna(brand['product_types']) else ''
    product_breadth = min(100, len(product_types.split(',')) * 20) if product_types else 0
    
    # Location prestige (simplified)
    location = brand['location'] if pd.notna(brand['location']) else ''
    location_prestige = 90 if location in ['UK', 'US', 'Germany', 'France', 'Italy', 'Spain'] else 70
    
    # Calculate weighted quality score
    quality_score = (0.3 * normalized_revenue) + \
                     (0.3 * deals_prestige) + \
                     (0.2 * product_breadth) + \
                     (0.2 * location_prestige)
    
    return max(0, min(100, quality_score))
# Sports Clothing Brand quality calculation (continued)
def calculate_clothing_brand_quality(brand, brands_df):
    """Calculate quality score for a sports clothing brand"""
    
    # Annual revenue (simplified - parse from string and normalize)
    def extract_revenue(revenue_str):
        if not pd.notna(revenue_str):
            return 50
        try:
            # Extract first number from string
            import re
            matches = re.findall(r'\d+', revenue_str)
            if matches:
                return int(matches[0])
            return 50
        except:
            return 50
    
    revenues = brands_df['annual_revenue'].apply(extract_revenue)
    max_revenue = revenues.max()
    min_revenue = revenues.min()
    brand_revenue = extract_revenue(brand['annual_revenue'])
    normalized_revenue = normalize_value(brand_revenue, min_revenue, max_revenue)
    
    # Sponsorship deals prestige (simplified - count number of deals)
    sponsorship_deals = brand['sponsorship_deals'] if pd.notna(brand['sponsorship_deals']) else ''
    deals_prestige = min(100, len(sponsorship_deals.split(',')) * 15) if sponsorship_deals else 0
    
    # Product types breadth (number of product types)
    product_types = brand['product_types'] if pd.notna(brand['product_types']) else ''
    product_breadth = min(100, len(product_types.split(',')) * 20) if product_types else 0
    
    # Location prestige (simplified)
    location = brand['location'] if pd.notna(brand['location']) else ''
    location_prestige = 90 if location in ['UK', 'US', 'Germany', 'France', 'Italy', 'Spain'] else 70
    
    # Calculate weighted quality score
    quality_score = (0.3 * normalized_revenue) + \
                     (0.3 * deals_prestige) + \
                     (0.2 * product_breadth) + \
                     (0.2 * location_prestige)
    
    return max(0, min(100, quality_score))
# Traveling Agency quality calculation
def calculate_traveling_agency_quality(agency, agencies_df):
    """Calculate quality score for a traveling agency"""
    
    # Services offered breadth
    services = agency['services_offered'] if pd.notna(agency['services_offered']) else ''
    services_breadth = min(100, len(services.split(',')) * 20) if services else 0
    
    # Affiliated clients prestige
    affiliated_clients = agency['affiliated_clients'] if pd.notna(agency['affiliated_clients']) else ''
    clients_prestige = min(100, len(affiliated_clients.split(',')) * 10) if affiliated_clients else 0
    
    # Operating regions breadth
    regions = agency['operating_regions'] if pd.notna(agency['operating_regions']) else ''
    regions_breadth = min(100, len(regions.split(',')) * 20) if regions else 0
    
    # Travel partners prestige
    partners = agency['travel_partners'] if pd.notna(agency['travel_partners']) else ''
    partners_prestige = min(100, len(partners.split(',')) * 20) if partners else 0
    
    # Calculate weighted quality score
    quality_score = (0.3 * services_breadth) + \
                     (0.3 * clients_prestige) + \
                     (0.2 * regions_breadth) + \
                     (0.2 * partners_prestige)
    
    return max(0, min(100, quality_score))

# Service Provider quality calculation
def calculate_service_provider_quality(provider, providers_df):
    """Calculate quality score for a service provider"""
    
    # Service type value (simplified)
    service_type = provider['service_type'] if pd.notna(provider['service_type']) else ''
    service_type_value = 90 if service_type in ['Medical', 'Legal', 'Financial', 'Training'] else 70
    
    # Partnerships made prestige
    partnerships = provider['partnerships_made'] if pd.notna(provider['partnerships_made']) else ''
    partnerships_prestige = min(100, len(partnerships.split(',')) * 15) if partnerships else 0
    
    # Regional coverage (simplified)
    region = provider['region'] if pd.notna(provider['region']) else ''
    regional_coverage = 90 if region in ['Europe', 'North America', 'Global'] else 70
    
    # Calculate weighted quality score
    quality_score = (0.3 * service_type_value) + \
                     (0.3 * partnerships_prestige) + \
                     (0.2 * regional_coverage) + \
                     (0.2 * 0)  # No verified status in dataset, so 0 bonus
    
    return max(0, min(100, quality_score))

# Communication Box quality calculation (similar to service provider)
def calculate_communication_box_quality(box, boxes_df):
    """Calculate quality score for a communication box"""
    
    # Services offered breadth
    services = box['services_offered'] if pd.notna(box['services_offered']) else ''
    services_breadth = min(100, len(services.split(',')) * 20) if services else 0
    
    # Affiliated clients prestige
    affiliated_clients = box['affiliated_clients'] if pd.notna(box['affiliated_clients']) else ''
    clients_prestige = min(100, len(affiliated_clients.split(',')) * 10) if affiliated_clients else 0
    
    # Partnerships made prestige
    partnerships = box['partnerships_made'] if pd.notna(box['partnerships_made']) else ''
    partnerships_prestige = min(100, len(partnerships.split(',')) * 15) if partnerships else 0
    
    # Regional coverage (simplified)
    region = box['region'] if pd.notna(box['region']) else ''
    regional_coverage = 90 if region in ['Europe', 'North America', 'Global'] else 70
    
    # Calculate weighted quality score
    quality_score = (0.3 * services_breadth) + \
                     (0.3 * clients_prestige) + \
                     (0.2 * partnerships_prestige) + \
                     (0.2 * regional_coverage)
    
    return max(0, min(100, quality_score))

```

```python
def calculate_entity_quality(entity_id, entity_type, datasets):
    """
    Calculate quality score for an entity based on its type.
    
    Parameters:
    - entity_id: ID of the entity
    - entity_type: Type of entity (player, club, etc.)
    - datasets: Dictionary containing all entity datasets
    
    Returns:
    - Quality score (0-100)
    """
    try:
        if entity_type == 'player':
            df = datasets['players']
            entity = df[df['player_id'] == entity_id].iloc[0]
            return calculate_player_quality(entity, df)
            
        elif entity_type == 'club':
            df = datasets['clubs']
            entity = df[df['id'] == entity_id].iloc[0]
            return calculate_club_quality(entity, df)
            
        elif entity_type == 'player_agent':
            df = datasets['player_agents']
            entity = df[df['license_number'] == entity_id].iloc[0]
            return calculate_player_agent_quality(entity, df)
            
        elif entity_type == 'recruiting_agent':
            df = datasets['recruiting_agents']
            entity = df[df['scout_id'] == entity_id].iloc[0]
            return calculate_recruiting_agent_quality(entity, df)
            
        elif entity_type == 'manager':
            df = datasets['managers_staff']
            entity = df[df['id'] == entity_id].iloc[0]
            return calculate_manager_quality(entity, df)
            
        elif entity_type == 'sporting_mgmt':
            df = datasets['sporting_mgmt']
            entity = df[df['agency_id'] == entity_id].iloc[0]
            return calculate_sporting_mgmt_quality(entity, df)
            
        elif entity_type == 'fitness_club':
            df = datasets['fitness_clubs']
            entity = df[df['club_id'] == entity_id].iloc[0]
            return calculate_fitness_club_quality(entity, df)
            
        elif entity_type == 'equipment_supplier':
            df = datasets['equipment_suppliers']
            entity = df[df['supplier_id'] == entity_id].iloc[0]
            return calculate_equipment_supplier_quality(entity, df)
            
        elif entity_type == 'clothing_brand':
            df = datasets['clothing_brands']
            entity = df[df['brand_id'] == entity_id].iloc[0]
            return calculate_clothing_brand_quality(entity, df)
            
        elif entity_type == 'sponsor':
            df = datasets['sponsors']
            entity = df[df['sponsor_id'] == entity_id].iloc[0]
            return calculate_sponsor_quality(entity, df)
            
        elif entity_type == 'traveling_agency':
            df = datasets['traveling_agencies']
            entity = df[df['agency_id'] == entity_id].iloc[0]
            return calculate_traveling_agency_quality(entity, df)
            
        elif entity_type == 'service_provider':
            df = datasets['service_providers']
            entity = df[df['provider_id'] == entity_id].iloc[0]
            return calculate_service_provider_quality(entity, df)
            
        elif entity_type == 'communication_box':
            df = datasets['communication_boxes']
            entity = df[df['box_id'] == entity_id].iloc[0]
            return calculate_communication_box_quality(entity, df)
            
        else:
            # Unknown entity type
            print(f"Warning: Unknown entity type '{entity_type}'")
            return 50  # Default score
            
    except (IndexError, KeyError) as e:
        # If entity not found or another error occurs, return default score
        print(f"Warning: Could not calculate quality for {entity_type} with ID {entity_id}. Error: {str(e)}")
        return 50
```

```python
def determine_verification_status(entity_type, score):
    """
    Determine if an entity should be verified based on its score.
    
    Parameters:
    - entity_type: Type of entity (player, club, etc.)
    - score: Verification score
    
    Returns:
    - Boolean indicating verification status
    """
    # Define thresholds for each entity type
    thresholds = {
        'player': 150,
        'club': 200,
        'player_agent': 180,
        'recruiting_agent': 180,
        'manager': 160,
        'sporting_mgmt': 170,
        'fitness_club': 150,
        'equipment_supplier': 150,
        'clothing_brand': 150,
        'sponsor': 150,
        'traveling_agency': 140,
        'service_provider': 140,
        'communication_box': 140
    }
    
    # Get threshold for this entity type, default to 150
    threshold = thresholds.get(entity_type, 150)
    
    # Return verification status
    return score >= threshold
```

```python
def calculate_verification_score(entity_id, entity_type, likes_df, base_weights, datasets):
    """
    Calculate verification score for an entity based on likes from other entities.
    
    Parameters:
    - entity_id: ID of the entity to verify
    - entity_type: Type of entity (player, club, etc.)
    - likes_df: DataFrame with likes data
    - base_weights: Dictionary of base weights
    - datasets: Dictionary containing all entity datasets
    
    Returns:
    - Tuple of (total_score, score_components)
    """
    # Get likes where this entity is liked
    entity_likes = likes_df[(likes_df['liked_id'] == entity_id) & (likes_df['liked_type'] == entity_type)]
    
    # Initialize score components
    score_components = []
    total_score = 0
    
    # No likes, return minimal score
    if len(entity_likes) == 0:
        return 0, score_components
    
    # Process each like
    for _, like in entity_likes.iterrows():
        liker_id = like['liker_id']
        liker_type = like['liker_type']
        
        # Get base weight for this liker type
        weight_matrix = base_weights.get(entity_type, {})
        base_weight = weight_matrix.get(liker_type, 1)  # Default to 1 if not defined
        
        # Calculate liker quality (0-100 scale)
        liker_quality = calculate_entity_quality(liker_id, liker_type, datasets) / 100.0
        
        # Calculate score contribution
        score_contribution = base_weight * liker_quality
        
        # Add to total score
        total_score += score_contribution
        
        # Add to components
        score_components.append({
            'liker_id': liker_id,
            'liker_type': liker_type,
            'base_weight': base_weight,
            'liker_quality': liker_quality * 100,  # Convert back to 0-100 scale for readability
            'score_contribution': score_contribution
        })
    
    return total_score, score_components
```

```python
def verify_entity(entity_id, entity_type, likes_df, base_weights, datasets):
    """
    Run verification process for a single entity.
    
    Parameters:
    - entity_id: ID of the entity to verify
    - entity_type: Type of entity (player, club, etc.)
    - likes_df: DataFrame with likes data
    - base_weights: Dictionary of base weights
    - datasets: Dictionary of all datasets
    
    Returns:
    - Dictionary with verification results
    """
    # Calculate verification score
    score, components = calculate_verification_score(entity_id, entity_type, likes_df, base_weights, datasets)
    
    # Determine verification status
    verified = determine_verification_status(entity_type, score)
    
    # Get entity name
    try:
        if entity_type == 'player':
            name = datasets['players'][datasets['players']['player_id'] == entity_id]['full_name'].iloc[0]
        elif entity_type == 'club':
            name = datasets['clubs'][datasets['clubs']['id'] == entity_id]['club name'].iloc[0]
        # Add other entity types as needed
        else:
            name = f"{entity_type}_{entity_id}"
    except:
        name = f"{entity_type}_{entity_id}"
        
    # Create result dictionary
    result = {
        'entity_id': entity_id,
        'entity_type': entity_type,
        'entity_name': name,
        'verification_score': score,
        'verified': verified,
        'score_components': components
    }
    
    return result
```
